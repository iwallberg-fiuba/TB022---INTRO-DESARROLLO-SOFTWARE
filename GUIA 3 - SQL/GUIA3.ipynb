## Guía 3 - SQL
 
 ## Introducción
 
 En esta guía, trabajaremos con SQL para realizar consultas sobre una base de datos ficticia. La base de datos contiene información sobre bandas musicales, álbumes, canciones y conciertos. A continuación, se presentan las tablas y sus relaciones:


```sql

-- Tabla de bandas musicales
CREATE TABLE bandas (
    id SERIAL PRIMARY KEY,
    nombre TEXT NOT NULL,
    pais_origen TEXT NOT NULL,
    fecha_creacion INTEGER NOT NULL,
    genero TEXT NOT NULL,
    cant_integrantes INTEGER NOT NULL
);

-- Tabla de álbumes
CREATE TABLE albumes (
    id SERIAL PRIMARY KEY,
    banda_id INTEGER NOT NULL REFERENCES bandas(id),
    nombre TEXT NOT NULL,
    lanzamiento INTEGER NOT NULL,
    duracion INTEGER NOT NULL,  -- duración en minutos
    ranking INTEGER UNIQUE NOT NULL  -- ranking global entre todos los álbumes
);

-- Tabla de canciones
CREATE TABLE canciones (
    id SERIAL PRIMARY KEY,
    banda_id INTEGER NOT NULL REFERENCES bandas(id),
    album_id INTEGER NOT NULL REFERENCES albumes(id),
    nombre TEXT NOT NULL,
    duracion INTEGER NOT NULL,  -- duración en minutos
    ranking INTEGER UNIQUE NOT NULL  -- ranking global entre todas las canciones
);

-- Tabla de conciertos
CREATE TABLE conciertos (
    id SERIAL PRIMARY KEY,
    nombre TEXT NOT NULL,
    pais TEXT NOT NULL,
    fecha INTEGER NOT NULL  -- año
);

-- Tabla intermedia para relación muchos-a-muchos entre conciertos y músicos
CREATE TABLE conciertos_musicos (
    concierto_id INTEGER NOT NULL REFERENCES conciertos(id),
    banda_id INTEGER NOT NULL REFERENCES bandas(id),
    PRIMARY KEY (concierto_id, banda_id)
);
```

#### Ejercicio 2

El nombre de las canciones (con el nombre de su álbum) que pertenecen a álbumes lanzados después del año 2000. Ordenarlas por nombre de álbum (de A a Z), y por cada álbum, por nombre de canción (de Z a A).


**Idea:**

- Se unen `canciones` y `albumes`.
- Se filtran álbumes lanzados después de 2000.
- Se ordena según la consigna.

````python
SELECT canciones.nombre, albumes.nombre FROM canciones
JOIN albumes
    ON albumes.id = canciones.album_id
WHERE albumes.lanzamiento > 2000
ORDER BY albumes.nombre ASC, canciones.nombre DESC;
````

#### Ejercicio 3

El nombre de las bandas con menos de 5 integrantes y de género "Rock Alternativo", cuyas canciones duran 4 o más minutos (todas ellas).

- Se filtran bandas con menos de 5 integrantes y género `Rock Alternativo`.

````python
SELECT bandas.nombre FROM bandas
WHERE bandas.cant_integrantes < 5 AND bandas.genero = 'Rock Alternativo'
AND NOT EXISTS (
    SELECT 1 FROM canciones
    WHERE canciones.banda_id = bandas.id AND canciones.duracion < 4
);
````

#### Ejercicio 4

El concierto con más bandas participantes, mostrando el nombre del concierto y la cantidad de bandas que participaron.


**Idea:**

- Se cuentan las bandas por concierto.
- Se ordena de mayor a menor cantidad.
- Se toma el primer resultado.

````python
SELECT conciertos.nombre, COUNT(*) AS cantidad_bandas FROM conciertos
JOIN conciertos_musicos
    ON conciertos.id = conciertos_musicos.concierto_id
GROUP BY conciertos.id
ORDER BY cantidad_bandas DESC
LIMIT 1;
````

#### Ejercicio 5

El nombre de los álbumes cuyas bandas fueron al menos a un concierto en el país "Argentina", ordenados por ranking de mejor a peor.


**Idea:**

- Se filtran los álbumes cuyas bandas aparecen en conciertos realizados en `Argentina`.
- Se ordena por `ranking` de mejor a peor.

````python
SELECT a.nombre
FROM albumes a
WHERE a.banda_id IN (
    SELECT DISTINCT cm.banda_id
    FROM conciertos_musicos cm
    JOIN conciertos c ON cm.concierto_id = c.id
    WHERE c.pais = 'Argentina'
)
ORDER BY a.ranking ASC;
````

La cantidad de canciones que tiene cada banda, ordenadas de mayor a menor cantidad de canciones. Mostrar el nombre de la banda y la cantidad de canciones.

### Nivel 4 (Uy, qué está pasando?)


#### Ejercicio 1

El primer álbum de todas las bandas. Mostrar el nombre de la banda, del álbum y el año de lanzamiento. ¿Y para hacerlo con el último de cada banda?


**Idea:**

- La subconsulta usa `MIN` para obtener el primer álbum de cada banda.
- Para obtener el último, se reemplazaría `MIN` por `MAX`.

````python
SELECT bandas.nombre AS banda, albumes.nombre AS album, albumes.lanzamiento
FROM albumes
JOIN bandas
    ON bandas.id = albumes.banda_id
WHERE albumes.lanzamiento = (
    SELECT MIN(albumes2.lanzamiento) FROM albumes AS albumes2
    WHERE albumes2.banda_id = albumes.banda_id
);
````

#### Ejercicio 2

La cantidad promedio de integrantes de las bandas por cada género musical. Mostrar el género y la cantidad promedio de integrantes.


**Idea:**

- Se calcula el promedio de integrantes por género.
- Se redondea el resultado a 2 decimales.

````python
SELECT bandas.genero, ROUND(AVG(bandas.cant_integrantes), 2) AS promedio_integrantes FROM bandas
GROUP BY bandas.genero;
````

#### Ejercicio 3

El nombre de las bandas que participaron en todos los conciertos que ocurrieron en el país "Argentina".


**Idea:**

- Se usa un doble `NOT EXISTS` para verificar que no falte ningún concierto de `Argentina` por banda.

````python
SELECT bandas.nombre FROM bandas
WHERE NOT EXISTS (
    SELECT 1 FROM conciertos
    WHERE conciertos.pais = 'Argentina'
        AND NOT EXISTS (
            SELECT 1 FROM conciertos_musicos
            WHERE conciertos_musicos.concierto_id = conciertos.id
                AND conciertos_musicos.banda_id = bandas.id
        )
);
````

#### Ejercicio 4

Las bandas cuyo promedio de duración de canciones es mayor a 5 minutos. Mostrar el nombre de la banda y el promedio de duración de sus canciones.


**Idea:**

- Se calcula el promedio de duración de canciones por banda.
- Se filtran los promedios mayores a 5 minutos.

````python
SELECT bandas.nombre, ROUND(AVG(canciones.duracion), 2) AS promedio_duracion FROM bandas
JOIN canciones
    ON bandas.id = canciones.banda_id
GROUP BY bandas.id
HAVING AVG(canciones.duracion) > 5;
````

#### Ejercicio 5

El nombre de las bandas que no tienen conciertos registrados.


**Idea:**

- Se excluyen las bandas cuyos IDs aparecen en `conciertos_musicos` mediante `NOT IN`.

````python
SELECT b.nombre
FROM bandas b
WHERE b.id NOT IN (
    SELECT DISTINCT cm.banda_id
    FROM conciertos_musicos cm
);
````

#### Ejercicio 6

El nombre y ranking de los álbumes que tienen todas sus canciones con un ranking peor o igual a 30.


**Idea:**

- Se usan `NOT EXISTS` para descartar álbumes que tengan alguna canción con ranking 30 o mejor.

````python
SELECT a.nombre, a.ranking
FROM albumes a
WHERE NOT EXISTS (
    SELECT 1 FROM canciones c
    WHERE c.album_id = a.id AND c.ranking <= 30
);
````

#### Ejercicio 7

El nombre de las bandas que tienen a más de la mitad de sus canciones con un ranking peor o igual a 30. Mostrar también (para cada banda) la cantidad de canciones rankeadas peor o igual a 30, así como el promedio de ranking de TODAS las canciones de la banda.


**Idea:**

- Se cuentan las canciones con ranking mayor a 30 por banda.
- Se calcula el promedio de ranking de todas sus canciones.
- Se filtran las bandas donde esa cantidad supera a la mitad del total.

````python
SELECT bandas.nombre,
       COUNT(CASE WHEN canciones.ranking > 30 THEN 1 END) AS cantidad_rankeadas,
       ROUND(AVG(canciones.ranking), 2) AS promedio_ranking
FROM bandas
JOIN canciones ON bandas.id = canciones.banda_id
GROUP BY bandas.id
HAVING COUNT(CASE WHEN canciones.ranking > 30 THEN 1 END) > COUNT(*) / 2;
````

#### Ejercicio 8

El nombre de las bandas que no tienen ninguna canción dentro del TOP 10 de canciones (según fulanito). Mostrar también el promedio de ranking de todas las canciones de la banda, y ranking mínimo de ellas (es decir, el de su mejor canción).


**Idea:**

- Se calcula el promedio de ranking por banda.
- Se obtiene también el `ranking_minimo` de sus canciones.
- Se conservan solo las bandas cuyo mejor ranking es mayor a 10.

````python
SELECT bandas.nombre, ROUND(AVG(canciones.ranking), 2) AS promedio_ranking, MIN(canciones.ranking) AS ranking_minimo
FROM bandas
JOIN canciones
    ON bandas.id = canciones.banda_id
GROUP BY bandas.id
HAVING MIN(canciones.ranking) > 10;
````

#### Ejercicio 9

El nombre de las canciones cuyo ranking es mejor que el ranking del álbum al que pertenecen. Mostrar también el ranking de la canción, así como el nombre del álbum y su ranking. Ordenar por ranking de canción (de mejor a peor).


**Idea:**

- Se comparan los rankings de `canciones` y `albumes`.
- Se devuelven solo las canciones con mejor ranking que su álbum.
- Se ordena por ranking de canción.

````python
SELECT canciones.nombre AS cancion, canciones.ranking AS ranking_cancion, albumes.nombre AS album, albumes.ranking AS ranking_album
FROM canciones
JOIN albumes
    ON albumes.id = canciones.album_id
WHERE canciones.ranking < albumes.ranking
ORDER BY canciones.ranking ASC;
````

#### Ejercicio 10

Las bandas y los conciertos donde el concierto se realizó en el país de origen de la banda. Mostrar (en este orden) el nombre de la banda, el del concierto, y el país de origen de la banda.


**Idea:**

- Se unen bandas y conciertos a través de `conciertos_musicos`.
- Se filtran los casos donde el país del concierto coincide con el país de origen de la banda.

````python
SELECT bandas.nombre, conciertos.nombre, bandas.pais_origen FROM bandas
JOIN conciertos_musicos
    ON bandas.id = conciertos_musicos.banda_id
JOIN conciertos
    ON conciertos.id = conciertos_musicos.concierto_id
WHERE conciertos.pais = bandas.pais_origen;
````
