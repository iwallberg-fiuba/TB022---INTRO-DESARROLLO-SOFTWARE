SQL:
Structured Query Language. Lenguaje para gesionar bases de datos relacionales.
Permite:
- definir estructuras
- manipular datos
- consultar informacion


- SELECT: Extrae datos de una o mas tablas
- FROM: Especifica la tabla de donde se extraeran los datos
- WHERE: Filtra los resultados segun una condicion
- ORDER BY: elegis una variable y pones ASC o DESC. Podes tener varias seguidas.
- LIMIT: limita la cantidad de resultados a mostrar

Siempre respetar el orden. SELECT -> FROM -> WHERE -> ORDER BY (opcional y puede repetirse). -> LIMIT (opcional)

```python
SELECT nombre, apellido FROM clientes WHERE ciudad = `Madrid` ORDER BY apellido DESC LIMIT 10;
```
