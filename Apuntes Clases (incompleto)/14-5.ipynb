## HTML y CSS

| Tecnología | Significado | Creador | Año | Objetivo |
|---|---|---|---|---|
| HTML | HyperText Markup Language | Tim Berners-Lee | 1991 | Estructurar contenido web |
| CSS | Cascade Style Sheets | Håkon Wium Lie | 1994 | Separar estructura y diseño |

- HTML nació en el CERN para conectar información mediante hipertexto.
- CSS apareció para resolver el caos visual de la web primitiva.

## Conceptos generales

- Tag = etiqueta
- Las etiquetas normalmente:
  - se abren y cierran
  - pueden contener otras etiquetas
- Excepciones:
  - `<br>`
  - `<hr>`
  - `<meta>`
- HTML funciona como un árbol jerárquico.
- Markdown:
  ```md
  # Título
```

equivale a:

```html
<h1>Título</h1>
```

## `div`

### Para qué sirve

`div` es un contenedor genérico usado para:

* agrupar elementos
* crear layouts
* estructurar la UI
* aplicar estilos
* organizar componentes

### Qué puede contener

* títulos
* párrafos
* imágenes
* videos
* botones
* formularios
* listas
* otros `div`

```html
<div>
    <h1>Título</h1>
    <p>Párrafo</p>
    <button>Enviar</button>
</div>
```

### Cuándo NO usar `div`

Si el contenido tiene significado semántico, conviene usar etiquetas específicas.

| En vez de              | Mejor usar |
| ---------------------- | ---------- |
| `<div>menu</div>`      | `<nav>`    |
| `<div>contenido</div>` | `<main>`   |

## `nav`

`nav` viene de *navigation*.

Se usa para:

* menús
* barras de navegación
* links importantes

```html
<nav>
    <a href="/">Inicio</a>
    <a href="/productos">Productos</a>
</nav>
```

## Atributos comunes

| Atributo      | Función             |
| ------------- | ------------------- |
| `placeholder` | Texto temporal      |
| `value`       | Valor inicial       |
| `required`    | Campo obligatorio   |
| `disabled`    | Deshabilita         |
| `checked`     | Marcado por defecto |
| `min` / `max` | Límites             |
| `id`          | Identificador único |
| `class`       | Clase reutilizable  |

### `id`

* No debería repetirse.
* CSS usa `#`.

```html
<input id="usuario">
```

```css
#usuario {
    border: 1px solid red;
}
```

```javascript
document.getElementById("usuario")
```

---

### `class`

* Puede repetirse.
* CSS usa `.`.

```html
<input class="input-grande">
```

```css
.input-grande {
    font-size: 20px;
}
```

```javascript
document.querySelector(".input-grande")
```

---

## Diferencia entre `id` y `class`

| Atributo | Se repite | CSS      |
| -------- | --------- | -------- |
| `id`     | No        | `#id`    |
| `class`  | Sí        | `.class` |
