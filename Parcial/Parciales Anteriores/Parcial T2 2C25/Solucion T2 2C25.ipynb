### Ejercicio 5

El error se cometió en la etapa de diseño del sistema. En ese momento se deberían haber definido mejor la arquitectura del sistema, como una estructura escalable que permita mantener y extender el sistema de forma sencilla.

La decisión principal que se debería haber tomado es diseñar el sistema con separación de responsabilidades, dividiéndolo en módulos o componentes independientes. Esto facilita el mantenimiento, mejora la escalabilidad y reduce el acoplamiento entre las distintas partes del sistema.

Por ejemplo, en lugar de concentrar todo el código en un único módulo, se podría haber separado en un módulo de interfaz, uno de lógica de negocio y otro de acceso a datos. De esta forma, modificar una parte del sistema no afecta directamente a las demás.

### Ejercicio 6

Las ramas (branches) en Git son líneas independientes de desarrollo dentro de un repositorio. Permiten trabajar en nuevas funcionalidades, correcciones o experimentos sin afectar la versión principal del proyecto.

Cada rama mantiene su propio conjunto de cambios, que luego pueden integrarse al proyecto principal mediante operaciones como merge. Esto facilita el trabajo en equipo y permite desarrollar en paralelo sin generar conflictos por pisar el código de los demás o duplicarlo accidentalmente.

Por ejemplo, se puede tener una rama principal como `main` con la versión estable del sistema, una rama `feature-login` para desarrollar una nueva funcionalidad y una rama `fix-bug` para corregir errores específicos.

### Ejercicio 7

Parte a)  
- `INNER JOIN`: combina filas de dos tablas cuando existe coincidencia según la condición del JOIN (no necesariamente solo con primary y foreign keys). Excluye las filas sin coincidencia.
- `LEFT JOIN`: devuelve todas las filas de la tabla de la izquierda y las coincidentes de la tabla de la derecha. Si no hay coincidencia, los campos de la tabla derecha aparecen como `NULL`.
- `RIGHT JOIN`: devuelve todas las filas de la tabla de la derecha y las coincidentes de la tabla de la izquierda. Si no hay coincidencia, los campos de la tabla izquierda aparecen como `NULL`.

Parte c)  
Que una columna sea FOREIGN KEY a otra tabla significa que los valores que se almacenan en esa columna deben coincidir con valores existentes de la PRIMARY KEY (u otra columna única) de la tabla a la que se hace referencia.

Esto se utiliza para relacionar tablas entre sí y asegurar la integridad referencial. Es decir, evita que se puedan guardar valores que no existan en la tabla relacionada.

Por ejemplo, si tenemos una tabla artistas con una columna id como clave primaria, y otra tabla albumes con una columna artista_id como clave foránea, entonces cada valor de artista_id debe corresponder a un id válido de la tabla artistas. No se podría insertar un álbum con un artista_id que no exista en la tabla de artistas.
