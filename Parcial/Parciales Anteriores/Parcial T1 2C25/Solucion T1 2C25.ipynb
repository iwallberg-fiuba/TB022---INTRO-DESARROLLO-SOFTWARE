### Ejercicio 1

Parte a)
1. `mkdir curso`: crea el directorio curso. Si ya existe, muestra un error.
2. `mv apunte.txt curso`: mueve el archivo apunte.txt al directorio curso. Si el archivo no existe, muestra un error.
3. `cd curso`: cambia al directorio curso. Si no existe, muestra un error.
4. `cat apunte.txt`: muestra en la terminal el contenido de apunte.txt (dentro de curso).
5. `echo "Linux es poderoso" > apunte.txt`: escribe ese texto en apunte.txt, sobrescribiendo su contenido anterior o creando el archivo si no existe.

Parte b)
Si se ejecuta cat apunte.txt después de los comandos anteriores (y todos se ejecutaron correctamente), el archivo únicamente mostrará la frase "Linux es poderoso". Esto se debe a que el operador `>` sobrescribe el contenido previo del archivo.

### Ejercicio 2


- `docker-compose build`  
  Construye las imágenes definidas en el archivo `docker-compose.yml`.

- `docker-compose up`  
  Crea y ejecuta los contenedores.

- `docker-compose ps`  
  Muestra el estado de los contenedores (en ejecución, detenidos, etc.).

- `docker-compose logs`  
  Muestra los logs generados por los contenedores.

- `docker-compose down`  
  Detiene y elimina los contenedores.

**Orden correcto:** 

Preparación:
- `docker-compose build`

Ejecución:
- `docker-compose up`

Verificación:
- `docker-compose ps`
- `docker-compose logs`

Finalización:
- `docker-compose down`

### Ejercicio 6
El comando git push se encarga de enviar los cambios realizados en el repositorio local hacia un repositorio remoto, como GitHub.
El comando git pull se utiliza para traer al repositorio local los cambios que se hicieron en el repositorio remoto, por ejemplo desde otras máquinas.

Un repositorio es un directorio que está siendo gestionado por Git para llevar el control de versiones de sus archivos. Esto implica que contiene una carpeta oculta .git, donde se guarda toda la información del historial y configuración.

Los repositorios pueden ser locales, es decir, existir únicamente en una máquina, o remotos, alojados en servicios como GitHub. En este esquema, Git se encarga del versionado local, mientras que el repositorio remoto actúa como una copia compartida en la nube.

Parte a)
```sql
SELECT DISTINCT albumes.nombre FROM albumes
JOIN cantantes
ON cantantes.id = albumes.artista_id
WHERE albumes.duracion < 60 AND cantantes.nombre LIKE 'A%'
ORDER BY albumes.anio_lanzamiento DESC;
```
