```bash
#!/bin/bash

lado1=$1
lado2=$2
lado3=$3

# Pienso en todas las condiciones mínimas para que sea equilátero, isósceles o escaleno y descarto primero el de condiciones más fáciles.

if (( lado1 == lado2 && lado2 == lado3 )); then
    echo "Es equilátero."
elif (( lado1 == lado2 || lado1 == lado3 || lado2 == lado3 )); then
    echo "Es isósceles."
else
    echo "Es escaleno."
fi
```

#### 1.3 Iteraciones

1. Crear un script de Bash que imprima los primeros n números naturales. Solicitar al usuario el ingreso de n.

```bash
#!/bin/bash

read -r -p "Ingrese hasta qué n imprimir: " n

for ((i=1; i<=n; i++)); do
    echo "$i"
done
```

2. Crear un script de Bash que solicite el ingreso de un número n y que luego calcule su factorial.

```bash
#!/bin/bash

read -r -p "Ingrese un n para calcular su factorial: " n

resultado=1 # Para que se cumpla que 0! = 1.

for ((i=1; i<=n; i++)); do # si n=0, no entra.
    resultado=$((resultado * i))
done

echo "El factorial de $n es: $resultado"
```

3. Crear un script de Bash que solicite al usuario el ingreso de un número par. En caso de que sea impar, solicitarlo nuevamente.

```bash
#!/bin/bash

while true; do # infinito hasta break
    read -r -p "Ingrese un número par: " n
    resto=$((n % 2))

    if (( resto == 0 )); then 
        echo "$n es par" # si es par, termina el programa.
        break
    fi

    echo "Inválido. Inténtelo de nuevo."
done
```

4. Crear un script de Bash que solicite un número que representa un día
de la semana, imprimir por pantalla de que día se trata. En caso de que
el número esté fuera de rango, solicitarlo nuevamente.

```bash
#!/bin/bash

while true; do
    read -r -p "Ingrese un número (1-7): " n

    # Se asume que el usuario ingresa un número válido.
    # Solo se controla que esté dentro del rango pedido.
    if (( n < 1 || n > 7 )); then
        echo "Error: el número debe estar entre 1 y 7" # Pide otro número de nuevo, salteándose el break.
    else
        # Imprime el caso y después se rompe para terminar el programa.

        case $n in
            1) echo "Lunes" ;;
            2) echo "Martes" ;;
            3) echo "Miércoles" ;;
            4) echo "Jueves" ;;
            5) echo "Viernes" ;;
            6) echo "Sábado" ;;
            7) echo "Domingo" ;;
        esac
        break
    fi
done
```

6. Crear un script de Bash que solicite al usuario un número natural n e
imprima en un archivo llamado “primos.txt” todos los números primos
existentes hasta n.

```bash
#!/bin/bash

read -r -p "Escriba un numero natural: " numero

for (( j=2; j<=numero; j++ )); do
    es_primo=1 # bandera=true

    for (( i=2; i*i<=j; i++ )); do # por cada número (j) hasta $numero , se prueban divisores (i) hasta la raíz de j.
        if (( j % i == 0 )); then
            es_primo=0 # bandera=false
            break
        fi
    done

    if (( es_primo == 1 )); then
        echo "$j" >> primos.txt
    fi
done
```
