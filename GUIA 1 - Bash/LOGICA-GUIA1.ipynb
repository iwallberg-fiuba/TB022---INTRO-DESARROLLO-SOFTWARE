## 1. Ejecutar un script

Pasos básicos:

1. Crear el archivo y escribir `#!/bin/bash` en la primera línea.
2. Guardarlo.
3. Darle permiso de ejecución con `chmod +x script.sh`.
4. Ejecutarlo con `./script.sh`.

Si el script recibe parámetros, se escriben al final del comando del punto `4`, antes de hacer `Enter` y separado por espacios.

En general, `read` conviene usarlo como `read -r -p "mensaje" variable`:

- `-r` evita interpretar caracteres de escape.
- `-p` muestra el mensaje y lee el valor en una sola línea.

## 2. Variables e input

En Bash, los parámetros y los valores leídos con `read` se reciben como texto.

```bash
parametro=$1  # Se recibe como string; incluso si el usuario escribe un número.
```

Bash no trabaja con un casteo explícito como otros lenguajes. Por eso, la idea correcta es validar siempre antes de hacer cuentas o comparaciones numéricas.

Dos conceptos distintos:

- Validación de input: comprobar que lo ingresado realmente tenga el formato esperado. Lo correcto es hacerlo con Regex.
- Validación de rango: comprobar que ese valor esté dentro del intervalo que necesitado, por ejemplo entre 1 y 10.

Si no se valida antes el input, la operación puede fallar o producir un comportamiento inesperado.

Entonces, la diferencia importante es esta:

- Con Regex se valida el input, o sea, se comprueba que lo ingresado realmente sea un número. Es la forma más completa cuando querés revisar el formato de lo que escribió el usuario.
- Sin Regex, lo que normalmente se hace es validar rangos o condiciones numéricas con `if`, pero eso solo tiene sentido si el valor ya fue validado como número o si la consigna permite asumir que ya lo es.

Por eso, ambas validaciones no reemplazan lo mismo: una valida el formato del dato y la otra valida qué condición cumple ese dato.

Por ahora, se permite asumir input válido (Regex no va a ser necesario); pero a veces la consigna pide la validación de rango.

```bash
# Ejemplo de validación de rango sin validación de input
if [ "$numero" -ge 1 ] && [ "$numero" -le 10 ]; then
    echo "El número está dentro del rango pedido."
else
    echo "El número está fuera del rango pedido."
fi
```

Operadores de comparación numérica con `[ ]`:

- `-gt`: greater than.
- `-lt`: less than.
- `-eq`: equal to.
- `-ge`: greater than or equal to.
- `-le`: less than or equal to.

En expresiones aritméticas con `(( ))` aparecen, por ejemplo:

- `||`: or.
- `&&`: and.
- `==`: comparación.
- `%`: resto.
- `=`: asignación.

## 3. Funciones

```bash
# Uso de parámetros "sin nombre"
suma() {
    resultado=$(( $1 + $2 ))
    echo "Suma: $resultado"
}

# Acá se nombran
read -r -p "Ingrese un número entero: " num1
read -r -p "Ingrese otro número entero: " num2

suma "$num1" "$num2"
```

En Bash, los parámetros de una función no se escriben en la definición como en Python. Se reciben dentro de la función con `$1`, `$2`, etc.

Cómo se lee ese script:

1. Se define la función `suma`.
2. Se leen dos valores.
3. Se llama a `suma` pasándole las variables ingresadas por el usuario.

Es importante entenderlo bien, porque a la hora de escribir código, se suele pensar de forma inconsciente en escribir primero lo que va a verse primero.  

En el uso de funciones, esa estrategia no sirve: tenemos que poner la función arriba de todo.

```bash
# Forma estándar
if [ condicion_basica ]; then
    algo
fi

# Para Operaciones Aritméticas
if (( numero % 2 == 0 )); then 
    algo
fi

# Para negar una condición básica
if ! [ condicion_basica ]; then
    algo
fi

# Para negar una condición en Operaciones Aritméticas
if (( numero % 2 != 0 )); then
    algo
fi
```

### `for anidado`

#### Es un `for` dentro de otro `for`

Por el Ejercicio 1.4.6

A veces se necesitan dos recorridos anidados. En el ejemplo de números primos:

- el `for` externo recorre cada número `j` desde 2 (para no tener problemas con el 1) hasta el valor ingresado `numero`;
- el `for` interno prueba posibles divisores `i` para ese `j`. Como se deben probar muchos números para cada `j`, se usa el `for anidado` (el `for` dentro del otro).
- `j` no llega a compararse con su propio valor, porque el divisor que se va probando es `i`, no `j`. Además, `i` nunca llega hasta `j`: el `for` interno corta antes, cuando deja de cumplirse `i*i<=j`.

Solo hace falta probar divisores hasta la raíz cuadrada de `j`, por lo que explica el teorema de la simetría de factores. La idea del teorema es esta: si un número tiene un divisor grande, entonces necesariamente tiene otro divisor más chico antes de la raíz del número. Por eso, si no aparece ningún divisor hasta la raíz cuadrada del número, ya no hace falta seguir buscando más allá.

```bash
#!/bin/bash

read -r -p "Escriba un número natural: " numero

for (( j=2; j<=numero; j++ )); do
    es_primo=1

    for (( i=2; i*i<=j; i++ )); do
        if (( j % i == 0 )); then
            es_primo=0
            break
        fi
    done

    if (( es_primo == 1 )); then
        echo "$j" >> primos.txt
    fi
done
```

### `while`

Por el Ejercicio 1.3.4

```bash
# Repetir si el valor NO cumple la condición.
while true; do
    read -r -p "Ingrese un número mayor que 10: " numero

    if [ "$numero" -gt 10 ]; then
        echo "Valor válido." # Se pueden añadir condiciones
        break
    fi

    echo "El valor no cumple la condición. Intente otra vez." # se puede poner en else
done

# Repetir si el valor SI cumple la condición.
while true; do
    read -r -p "Ingrese un número mayor que 10: " numero

    if [ "$numero" -le 10 ]; then
        echo "El valor no cumple la condición."
        continue # saltea break
    fi

    echo "Se procesa el valor válido." # se puede poner en else para añadir acciones.
    break
done
```

## 5. Arrays e iteraciones

#### Ejercicio 1.4.6 modo difícil:

```bash
dnis=("$@")
```

Se vería de la forma:

```bash
dnis=(47000000 56000000 30000000)
```

Para ordenarlo con `sort`, primero debemos convertir el array en una lista de líneas (así funciona `sort`).

```bash
array_a_string_ordenado=$(printf "%s\n" "${dnis[@]}" | sort -n)
```

Qué pasa en esa línea:

1. `printf "%s\n" "${dnis[@]}"` imprime cada elemento del array en una línea distinta.
2. `|` envía esa salida a `sort`.
3. `sort -n` ordena numéricamente de menor a mayor.
4. `$(...)` guarda el resultado final en una variable (todavía no se va a imprimir en pantalla).

La pipeline (`|`) conecta la salida del comando de la izquierda con la entrada del comando de la derecha.  
Básicamente, no se imprime nada en pantalla todavía.

```bash
# Para imprimirlo en pantalla
echo "DNIs ordenados de menor a mayor: "
echo "$array_a_string_ordenado"
```

Se usan dos echo para que el título salga en una línea y la lista ordenada empiece debajo. Si no, el primer valor de la variable quedaría pegado al texto del primer `echo` y el output se vería desprolijo.

#### Ejercicio 1.4.6 modo fácil:

```bash
#!/bin/bash

echo "DNIs ordenados: "

for dni in "$@"; do
    echo "$dni"
done | sort -n
```

- `-n` indica ordenar numéricamente, de menor a mayor.  
- La pipeline `|` hace que la salida de un comando pase a ser la entrada de otro.  

Entonces:
- `sort` necesita recibir datos para poder ordenarlos. En este ejemplo, la pipeline es una forma de pasarle esos datos.
- Cada iteración del `for` (cada DNI) pasa a ser la entrada de `sort`.
- `sort` devuelve todo lo almacenado de forma ordenada una vez terminado el for.

## 6. Redirecciones y archivos

- `>` redirige la salida a un archivo. Si no existe, lo crea; si existe, lo sobrescribe.
- `>>` agrega contenido al final del archivo. Si no existe, lo crea.
- `pwd` significa `print working directory`.
- `$USER` muestra el user del sistema. Útil.
- `echo` no puede imprimir contenido de archivos; pero sí imprimir *en* ellos.
- `cat` imprime en pantalla las líneas de un archivo.
- `ls`: lista los archivos del directorio actual
- `-1`: fuerza un archivo por línea
- `-S`: ordena archivos por tamaño, de mayor a menor
