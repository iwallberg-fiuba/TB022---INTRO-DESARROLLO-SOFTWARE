### Ejercicio 1

Un sistema Unix-like es un sistema operativo que, sin ser el Unix original, adopta su filosofía y forma de funcionamiento. Unix fue un sistema desarrollado en los años 70, mientras que Linux es un sistema posterior que fue creado desde cero inspirándose en Unix, por eso se lo conoce como "Unix-like".

Estos sistemas comparten características como ser multiusuario, multitarea, tener una estructura de archivos unificada y ofrecer la posibilidad de utilizar múltiples herramientas combinables desde la línea de comandos (por ejemplo, mediante `pipes`). En este sentido, Linux (y también macOS) siguen el modelo de Unix, aunque Linux no sea un derivado directo del Unix original.

La diferencia con Windows es que los sistemas Unix-like siguen una filosofía de diseño distinta: todo se trata como un archivo (incluyendo dispositivos), tienen un modelo de permisos simple y consistente (usuario, grupo y otros), y están fuertemente orientados al uso de la línea de comandos y la combinación de herramientas. Además, su estructura de archivos es única y jerárquica desde una raíz común (`/`). En cambio, Windows utiliza una estructura diferente (por ejemplo, múltiples unidades como `C:`), un modelo de permisos distinto y está más orientado al uso de interfaces gráficas.

### Ejercicio 2

Una imagen de Docker es una **plantilla inmutable** que contiene todo lo necesario para ejecutar una aplicación, incluyendo el código, las dependencias, configuraciones y el sistema base. A partir de una imagen se pueden crear múltiples contenedores idénticos, lo que permite reproducir el mismo entorno en distintas máquinas. A diferencia de lo que se podría pensar, la imagen no define servicios (eso corresponde a herramientas como Docker Compose) ni la persistencia de datos, ya que los volúmenes se configuran al momento de ejecutar el contenedor, por ejemplo mediante `docker run` o en un archivo `docker-compose.yml`. Es decir, la imagen sólo define el **entorno de ejecución** de la aplicación.

Un contenedor, en cambio, es una **instancia en ejecución de una imagen**. Es el entorno aislado donde la aplicación realmente corre, con sus propios procesos, red y sistema de archivos. Los contenedores son livianos, efímeros (pueden crearse y eliminarse fácilmente) y permiten ejecutar la misma aplicación de forma consistente en distintos entornos.

### Ejercicio 5

Parte a)

Estando en la rama `main`, la secuencia correcta de comandos sería:

1. Asegurarse de tener todo actualizado  
`git checkout development`  
`git pull origin development`  

2. Crear la feature branch desde development  
`git checkout -b feature-nueva`

3. Realizar cambios y subirlos  
`git add .`  
`git commit -m "Agrega nueva funcionalidad"`  
`git push origin feature-nueva`  

4. Volver a development y actualizar  
`git checkout development`  
`git pull origin development`  

5. Merge de la feature branch  
`git merge feature-nueva`  

6. Subir los cambios fusionados  
`git push origin development`  

7. (Opcional) eliminar la branch si ya no se usa  
`git branch -d feature-nueva`  
`git push origin --delete feature-nueva`  

  
Parte b)  
Fusionar las feature branches con `development` en lugar de hacerlo directamente con `main` es una buena práctica porque permite integrar y probar los cambios en un entorno intermedio antes de llevarlos a la versión estable. La rama `main` suele representar el código listo para producción, por lo que evitar cambios directos reduce el riesgo de introducir errores en una versión que debería ser confiable.

De esta forma, `development` actúa como un espacio de integración donde se combinan las nuevas funcionalidades, se detectan posibles conflictos y se realizan pruebas antes de pasar a `main`. Esto mejora la calidad del software, facilita el trabajo en equipo y permite mantener una versión estable mientras se continúa desarrollando en paralelo.

### Ejercicio 6

La Ingeniería de Software es la disciplina que se encarga de desarrollar software de forma organizada, aplicando métodos, herramientas y buenas prácticas para evitar hacerlo de manera improvisada. No se trata solo de programar, sino de todo el proceso: entender qué necesita el usuario, diseñar cómo va a funcionar el sistema, implementarlo, probarlo, corregir errores y mantenerlo a lo largo del tiempo.

Hoy en día es necesaria porque el software forma parte de casi todo: aplicaciones, bancos, sistemas de salud, redes sociales, etc. Estos sistemas son cada vez más grandes y complejos, por lo que no alcanza con escribir código sin planificación. Si no se aplican principios de Ingeniería de Software, es muy probable que el sistema tenga errores, sea difícil de modificar o incluso falle en momentos críticos. Por eso, esta disciplina ayuda a crear software más confiable, seguro, fácil de mantener y que pueda evolucionar con el tiempo, además de permitir que varias personas trabajen en conjunto de manera ordenada.

### Ejercicio 7

```sql
SELECT 
    equipos_campeon.nombre AS campeon,
    equipos_subcampeon.nombre AS subcampeon
FROM torneos
JOIN equipos AS equipos_campeon
    ON equipos_campeon.id = torneos.campeon_id
JOIN equipos AS equipos_subcampeon
    ON equipos_subcampeon.id = torneos.subcampeon_id
ORDER BY equipos_campeon.anio_de_fundacion ASC;
```

### Ejercicio 8

```sql
SELECT 
    torneos.nombre,
    equipos_campeon.nombre AS campeon,
    equipos_subcampeon.nombre AS subcampeon
FROM torneos
JOIN equipos AS equipos_campeon
    ON equipos_campeon.id = torneos.campeon_id
JOIN equipos AS equipos_subcampeon
    ON equipos_subcampeon.id = torneos.subcampeon_id
WHERE 
    equipos_campeon.anio_de_fundacion > torneos.anio
    OR equipos_subcampeon.anio_de_fundacion > torneos.anio;
```
