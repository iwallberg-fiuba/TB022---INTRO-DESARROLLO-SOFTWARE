### Ejercicio 1

Una imagen de Docker es una **plantilla inmutable** que contiene todo lo necesario para ejecutar una aplicación, incluyendo el código, las dependencias, configuraciones y el sistema base. A partir de una imagen se pueden crear múltiples contenedores idénticos, lo que permite reproducir el mismo entorno en distintas máquinas. A diferencia de lo que se podría pensar, la imagen no define servicios (eso corresponde a herramientas como Docker Compose) ni la persistencia de datos, ya que los volúmenes se configuran al momento de ejecutar el contenedor, por ejemplo mediante `docker run` o en un archivo `docker-compose.yml`. Es decir, la imagen sólo define el **entorno de ejecución** de la aplicación.

Un contenedor, en cambio, es una **instancia en ejecución de una imagen**. Es el entorno aislado donde la aplicación realmente corre, con sus propios procesos, red y sistema de archivos. Los contenedores son livianos, efímeros (pueden crearse y eliminarse fácilmente) y permiten ejecutar la misma aplicación de forma consistente en distintos entornos.

### Ejercicio 2

Un sistema Unix-like es un sistema operativo que, sin ser el Unix original, adopta su filosofía y forma de funcionamiento. Unix fue un sistema desarrollado en los años 70, mientras que Linux es un sistema posterior que fue creado desde cero inspirándose en Unix, por eso se lo conoce como "Unix-like".

Estos sistemas comparten características como ser multiusuario, multitarea, tener una estructura de archivos unificada y ofrecer la posibilidad de utilizar múltiples herramientas combinables desde la línea de comandos (por ejemplo, mediante `pipes`). En este sentido, Linux (y también macOS) siguen el modelo de Unix, aunque Linux no sea un derivado directo del Unix original.

La diferencia con Windows es que los sistemas Unix-like siguen una filosofía de diseño distinta: todo se trata como un archivo (incluyendo dispositivos), tienen un modelo de permisos simple y consistente (usuario, grupo y otros), y están fuertemente orientados al uso de la línea de comandos y la combinación de herramientas. Además, su estructura de archivos es única y jerárquica desde una raíz común (`/`). En cambio, Windows utiliza una estructura diferente (por ejemplo, múltiples unidades como `C:`), un modelo de permisos distinto y está más orientado al uso de interfaces gráficas.

### Ejercicio 5

Parte a)  
Para resolver conflictos de merge en Git se debe:

1. Cambiar a la rama donde querés integrar los cambios (por ejemplo, `main`).
2. Traer los cambios de la otra rama (`git merge nombre_rama`), lo que puede generar conflictos.
3. Consultar el estado actual con `git status` para identificar los archivos en conflicto.
4. Editar manualmente los archivos en conflicto, utilizando las marcas que deja Git (`<<<<<<<`, `=======`, `>>>>>>>`) para decidir qué cambios conservar.
5. Agregar los archivos resueltos al staging con `git add`.
6. Realizar el commit con `git commit -m "Resolución de conflictos"`.
7. Finalmente, subir los cambios con `git push`.

Parte b)  
Una `feature-branch` es una rama del proyecto en donde se trabaja una nueva funcionalidad específica. Permite trabajar de forma aislada sin afectar la rama principal (por ejemplo, `main`), integrando los cambios recién cuando la funcionalidad está completa.

Una rama es una línea de trabajo independiente dentro de un repositorio que permite organizar el desarrollo. Gracias al uso de ramas, múltiples desarrolladores pueden trabajar en paralelo sobre distintas partes del proyecto sin interferir entre sí, evitando sobrescribir cambios y reduciendo conflictos. Esto facilita la colaboración, el mantenimiento del código y una integración más controlada mediante operaciones como `merge`.
