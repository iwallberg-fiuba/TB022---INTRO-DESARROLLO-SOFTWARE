```python
Ejemplo de validación de input con Regex:

```

```python
numero=$1

if ! [[ "$numero" =~ ^[0-9]+$ ]]; then
    echo "El valor ingresado no es un número natural válido."
    exit 1
fi
```

Explicación: 
- `[[ ... ]]` se usa para evaluaciones más completas.
- `=~` compara el texto con una expresión regular.
- `^` marca el inicio del texto.
- `[0-9]` permite cualquier dígito del 0 al 9.
- `+` indica uno o más dígitos.
- `$` marca el final del texto.

Entonces, `^[0-9]+$` acepta únicamente una cadena formada por uno o más dígitos.

```python
Una vez validado que el input sí es numérico, ya se puede tratar como número y usarlo en operaciones aritméticas.
```
