## 2.1 - Ejercicios básicos

Archivo `oraciones.txt`:

```text
El perro corre rápido por el parque.
Marí­a compró manzanas, peras y bananas en el mercado.
La computadora tiene 16 GB de RAM y 512 GB de almacenamiento.
¡Qué día tan bonito para un paseo por la montaña!
¿Cuántas palabras puede contener esta oración?
El gato saltó sobre la mesa para atrapar el ratón.
Las clases de Intro comienzan a las 7:30 AM.
Juan y Martina juegan al fútbol todos los viernes.
```

#### 2.1.1 Las ocurrencias de la letra `s`

```regex
[sS]
```
```bash
grep -oi "s" oraciones.txt
```

#### 2.1.2 Las ocurrencias de un caracter seguido de la letra `o`

```regex
.[oO]
```
```bash
grep -oi ".o" oraciones.txt
```

#### 2.1.3 Todos los dígitos en el archivo

```regex
\d
```
```bash
grep -o "[0-9]" oraciones.txt
```

#### 2.1.4 Las ocurrencias del punto `.`

```regex
\.
```
```bash
grep -o "\." oraciones.txt
```

#### 2.1.5 Las ocurrencias de la letra `c` al comienzo de la palabra

```regex
\b[cC]
```
```bash
grep -oP "\b[cC]" oraciones.txt
```

#### 2.1.6 Las palabras que comienzan con la letra `p`

```regex
\bp\w*
```
```bash
grep -oP "\bp\w*" oraciones.txt
```

#### 2.1.7 ¿Qué devuelven las siguientes expresiones?

**a) `^L`** # Líneas que comienzan con "L".   
```text
La computadora tiene 16 GB de RAM y 512 GB de almacenamiento.
Las clases de Intro comienzan a las 7:30 AM.
```

**b) `!$`** # Líneas que terminan con "!".
```text
¡Qué tan bonito para un paseo por la montaña!
```

**c) `[ ]`** (espacio)
```text
un espacio por cada espacio presente en el archivo
```

**d) `[A-Z1-3]`** # Caracteres en rango [A-Z] (letras mayúsculas) OR [1-3] (alguno de esos números)
```text
E
M
L
1
G
B
2
1
2
A
M
...
```

## 2.2 - Ejercicios avanzados

Archivo `contraseñas.csv`:

```text
cool_cat123,BananaSplit99
xXx_panda_xXx,unicornu
spacetraveler42,42isTheAnswer!
flower_power,RedRoses
shadow_ninja,!ninjaWarrior22
_bubblegumqueen,candyLand2024
the_real_hero,p@ssw0rd!
pirateKing777,G0ld&Silver$$?
coffee_addict,latteLover123
bookworm89,openSesame!
pepeMaquina,Ave7#0
```

Se pide extraer el campo indicado sin incluir la coma.

#### 2.2.1 Los nombres de usuario que contienen números

```regex
^[^,]*\d[^,]*
```
```bash
grep -oP "^[^,]*\d[^,]*" contraseñas.csv
```
```text
cool_cat123
spacetraveler42
pirateKing777
bookworm89
```

#### 2.2.2 Los nombres de usuario con al menos un guión bajo `_`

```regex
^[^,]*_[^,]*
```
```bash
grep -oP "^[^,]*_[^,]*" contraseñas.csv
```
```text
cool_cat123
xXx_panda_xXx
flower_power
shadow_ninja
_bubblegumqueen
the_real_hero
coffee_addict
```

#### 2.2.3 Las contraseñas alfanuméricas

```regex
(?<=,)[A-Za-z0-9]+$
```
```bash
grep -oP "(?<=,)[A-Za-z0-9]+$" contraseñas.csv
```
```text
BananaSplit99
unicornu
RedRoses
candyLand2024
latteLover123
```

#### 2.2.4 Los nombres de usuario alfanuméricos

```regex
^[A-Za-z0-9]+(?=,)
```
```bash
grep -oP "^[A-Za-z0-9]+(?=,)" contraseñas.csv
```
```text
spacetraveler42
pirateKing777
bookworm89
pepeMaquina
```

#### 2.2.5 Las contraseñas que comienzan y terminan con el mismo caracter

```regex
(?<=,)(.).*\1$
```
```bash
grep -oP "(?<=,)(.).*\1$" contraseñas.csv
```
```text
unicornu
```

#### 2.2.6 Las contraseñas con exactamente 14 caracteres

```regex
(?<=,).{14}$
```
```bash
grep -oP "(?<=,).{14}$" contraseñas.csv
```
```text
42isTheAnswer!
G0ld&Silver$$?
```

#### 2.2.7 Las contraseñas seguras

```regex
(?<=,)(?=.*[a-z])(?=.*[A-Z])(?=.*\d)(?=.*[@$!%*?&#]).+$
```
```bash
grep -oP "(?<=,)(?=.*[a-z])(?=.*[A-Z])(?=.*\d)(?=.*[@$!%*?&#]).+$" contraseñas.csv
```
```text
42isTheAnswer!
!ninjaWarrior22
G0ld&Silver$$?
Ave7#0
```
