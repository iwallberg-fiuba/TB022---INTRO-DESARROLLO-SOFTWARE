## Enfoque general por temas

Para estudiar esta guia sin perderse, conviene pensarla en cuatro temas:

1. Fundamentos de lectura de regex.
2. Patrones para texto libre (archivo de oraciones).
3. Extraccion por campos en formato `usuario,contraseña`.
4. Validaciones compuestas (longitud, repeticion y reglas de seguridad).

La idea es relacionar cada regex con su objetivo tecnico, no con el numero del ejercicio.

Regla de lectura: de izquierda a derecha. Cada simbolo agrega una condicion nueva.

## Diccionario rapido

- Match: texto que cumple el patron.
- Consumir texto: la regex "agarra" caracteres y los mete en el resultado.
- No consumir texto: la regex solo revisa una condicion.
- Guardar una parte: poner `(...)` para recordar un pedacito de texto.
- Repetir lo guardado: usar `\1` para pedir "lo mismo que guarde antes".

Regla corta:

- Si el simbolo aparece en el resultado, consumio.
- Si solo sirve para validar contexto, no consumio.

### Ejemplo guiado: `\bp\w*`

- `\b` marca el borde de palabra.
- `p` obliga a que la palabra empiece con esa letra.
- `\w*` permite seguir con cero o mas caracteres de palabra.

Por eso esta regex sirve para encontrar palabras que empiezan con `p`, como `perro`, `parque` o incluso `p` sola.

## Tema 1: piezas base de una regex

| Familia | Simbolo | Para que sirve | Ejemplo |
| --- | --- | --- | --- |
| Comodin | `.` | cualquier caracter (salvo salto de linea) | `a.c` |
| Escape | `\.` | punto literal | `3\.14` |
| Clases rapidas | `\d`, `\w` | digito y caracter de palabra | `\d+`, `\w+` |
| Borde | `\b` | inicio/fin de palabra | `\bpan` |
| Repeticion | `*`, `+` | cero o mas / uno o mas | `\w*`, `\w+` |
| Anclajes | `^`, `$` | inicio y fin de linea | `^L`, `!$` |
| Guardar y repetir | `(...)`, `\1` | guardar un caracter y pedir el mismo al final | `(.)\1` |

Esta tabla sirve como mapa rapido: primero identificas la familia y despues ajustas el patron puntual.

## Tema 1.1: comandos grep y lectura de lookarounds

### Flags basicas
- `grep -o`: muestra solo la parte que matchea.
- `grep -i`: ignora mayusculas/minusculas.
- `grep -P`: habilita sintaxis PCRE (`\b`, lookarounds y `\1`).

### Lookarounds en una frase
- Lookahead `(?=...)`: mira lo que viene despues.
- Lookbehind `(?<=...)`: mira lo que hay antes.
- Ninguno de los dos "agarra" texto por si solo: solo validan contexto.

### Ejemplo corto 1
```bash
grep -oi "s" oraciones.txt
```
`-o` devuelve cada `s` individual.

### Ejemplo corto 2
```bash
grep -oP "(?<=,)[A-Za-z0-9]+$" contraseñas.csv
```
`(?<=,)` ubica el inicio despues de la coma y luego se extrae la contraseña.

## Tema 2: patrones para texto libre

En texto libre aparecen cuatro ideas principales.

### 1) Clases de caracteres
- `[sS]`: acepta `s` o `S`.
- `.[oO]`: cualquier caracter seguido de `o` u `O`.
- `[A-Z1-3]`: un caracter en ese rango, no una palabra completa.
- `[ ]`: el caracter espacio.

### 2) Digitos y comodines
- `\d`: un digito.
- `.`: cualquier caracter.
- `\.`: punto literal.

### 3) Bordes y anclajes
- `\b[cC]`: `c` o `C` al inicio de palabra.
- `^L`: linea que empieza con `L`.
- `!$`: linea que termina en `!`.

### 4) Inicio de palabra mas continuidad
- `\bp\w*`: palabra que empieza con `p` y puede continuar con cero o mas caracteres de palabra.

Patron mental recomendado: primero definir donde puede empezar el match, despues que caracteres acepta y finalmente como termina.

Despues de la tabla, va la lectura paso a paso de cada una.

### 2.1.1 `[sS]`
Que busca: la letra `s` en minuscula o mayuscula.

Lectura simple:
- `[sS]` acepta una `s` o una `S`.
- Es una clase de caracteres: se elige una opcion de las que estan adentro.

### 2.1.2 `.[oO]`
Que busca: un caracter cualquiera y despues una `o` o `O`.

Lectura simple:
- `.` puede ser casi cualquier caracter.
- `[oO]` obliga a que el segundo caracter sea `o` u `O`.
- Sirve para buscar pares de caracteres con una condicion fija.

### 2.1.3 `\d`
Que busca: digitos.

Lectura simple:
- `\d` significa "un numero".
- Es una forma corta de escribir lo mismo que `[0-9]`.

### 2.1.4 `\.`
Que busca: un punto real.

Lectura simple:
- `.` solo significa "cualquier caracter" cuando no esta escapado.
- `\.` cambia el significado y pide un punto literal.

### 2.1.5 `\b[cC]`
Que busca: una `c` o `C` al inicio de una palabra.

Lectura simple:
- `\b` marca el borde de palabra.
- `[cC]` pide una `c` o `C`.
- Juntos dicen: empezar justo al comienzo de una palabra y tomar esa letra.

### 2.1.6 `\bp\w*`
Que busca: palabras que empiezan con `p`.

Lectura simple:
- `\b` marca el borde de palabra.
- `p` obliga a que la palabra arranque con esa letra.
- `\w*` permite seguir con cero o mas caracteres de palabra.

### 2.1.7.1 `^L`
Que busca: lineas que empiezan con `L`.

Lectura simple:
- `^` significa inicio de linea.
- `L` obliga a que el primer caracter sea esa letra.

### 2.1.7.2 `!$`
Que busca: lineas que terminan con `!`.

Lectura simple:
- `$` marca el final de la linea.
- `!` obliga a que el ultimo caracter sea ese signo.

### 2.1.7.3 `[ ]`
Que busca: el caracter espacio.

Lectura simple:
- La clase `[ ]` contiene un espacio literal.
- Eso hace match con cada espacio que aparezca en el texto.

### 2.1.7.4 `[A-Z1-3]`
Que busca: una mayuscula entre `A` y `Z`, o un numero entre `1` y `3`.

Lectura simple:
- `[A-Z]` es el rango de letras mayusculas.
- `[1-3]` es el rango de esos numeros.
- Al estar juntos, la regex acepta cualquiera de esos caracteres, pero solo uno por match.

## Tema 3: extraccion por campos en CSV

Cuando cada linea tiene formato `usuario,contraseña`, el problema central es separar correctamente ambos campos.

```text
usuario,contraseña
```

Estrategias clave:

- `^[^,]*`: consume desde el inicio todo lo que no sea coma.
- `(?<=,)`: obliga a empezar justo despues de la coma sin incluirla.
- `(?=,)`: obliga a frenar justo antes de la coma sin incluirla.

Ejemplo de usuario con guion bajo:

```regex
^[^,]*_[^,]*
```

Lectura:

- `^`: inicio de linea.
- `[^,]*`: consumir hasta la coma.
- `_`: exigir al menos un guion bajo.
- `[^,]*`: seguir hasta antes de la coma.

Criterio practico: si queres extraer por posicion relativa a la coma, usa lookaround; si queres consumir de punta a punta, usa clases negadas como `[^,]`.

## Tema 3.5: explicacion regex por regex (parte 2.2)

Primero, una tabla corta para ubicar cada patron de un vistazo:

| Regex | Que filtra | Pista para recordarlo |
| --- | --- | --- |
| `^[^,]*\d[^,]*` | usuarios con numeros | buscar un digito dentro del usuario |
| `^[^,]*_[^,]*` | usuarios con `_` | buscar un guion bajo dentro del usuario |
| `(?<=,)[A-Za-z0-9]+$` | contraseñas solo alfanumericas | empezar despues de la coma y seguir hasta el final |
| `^[A-Za-z0-9]+(?=,)` | usuarios solo alfanumericos | empezar al inicio y frenar antes de la coma |
| `(?<=,)(.).*\1$` | contraseñas que empiezan y terminan igual | guardar el primero y pedir el mismo al final |
| `(?<=,).{14}$` | contraseñas de 14 caracteres | contar exactamente 14 caracteres despues de la coma |
| `(?<=,)(?=.*[a-z])(?=.*[A-Z])(?=.*\d)(?=.*[@$!%*?&#]).+$` | contraseñas seguras | revisar varias reglas antes de aceptar |

Despues de la tabla, va la lectura paso a paso de cada una.

### 2.2.1 `^[^,]*\d[^,]*`
Que busca: usuarios que tengan al menos un numero.

Lectura simple:
- `^` arranca al inicio de la linea.
- `[^,]*` avanza por el usuario sin pasar la coma.
- `\d` exige un digito en algun punto del usuario.
- `[^,]*` permite seguir hasta la coma.

### 2.2.2 `^[^,]*_[^,]*`
Que busca: usuarios con al menos un guion bajo `_`.

Lectura simple:
- `^` inicio de linea.
- `[^,]*` consume dentro del usuario.
- `_` obliga a que exista un guion bajo.
- `[^,]*` sigue hasta la coma.

### 2.2.3 `(?<=,)[A-Za-z0-9]+$`
Que busca: contraseñas solo alfanumericas.

Lectura simple:
- `(?<=,)` empieza justo despues de la coma.
- `[A-Za-z0-9]+` solo letras y numeros, uno o mas.
- `$` hasta fin de linea.

### 2.2.4 `^[A-Za-z0-9]+(?=,)`
Que busca: usuarios solo alfanumericos.

Lectura simple:
- `^` inicio de linea.
- `[A-Za-z0-9]+` letras o numeros del usuario.
- `(?=,)` frena justo antes de la coma.

### 2.2.5 `(?<=,)(.).*\1$`
Que busca: contraseñas que empiezan y terminan con el mismo caracter.

Lectura simple:
- `(?<=,)` empieza en la contraseña.
- `(.)` guarda el primer caracter.
- `.*` permite cualquier contenido en el medio.
- `\1` exige que el ultimo sea igual al primero guardado.
- `$` fin de linea.

### 2.2.6 `(?<=,).{14}$`
Que busca: contraseñas con longitud exacta 14.

Lectura simple:
- `(?<=,)` empieza despues de la coma.
- `.{14}` toma exactamente 14 caracteres.
- `$` termina justo al final.

### 2.2.7 `(?<=,)(?=.*[a-z])(?=.*[A-Z])(?=.*\d)(?=.*[@$!%*?&#]).+$`
Que busca: contraseñas "seguras" con varias reglas.

Reglas que exige:
- al menos una minuscula.
- al menos una mayuscula.
- al menos un numero.
- al menos un simbolo del conjunto `@$!%*?&#`.

Lectura simple:
- `(?<=,)` empieza en la contraseña.
- cada `(?=.*...)` chequea una regla.
- `.+$` toma la contraseña completa si cumple todas.

## Tema 4: lookarounds para extraer sin "comerse" separadores

Los lookarounds agregan condiciones alrededor del match, pero no consumen caracteres.

### Lookbehind positivo `(?<=...)`
Pide que antes del match exista un patron.

```regex
(?<=,)[A-Za-z0-9]+$
```

Sirve para extraer la contraseña: empieza despues de la coma y toma hasta el final.

### Lookahead positivo `(?=...)`
Pide que despues del match exista un patron.

```regex
^[A-Za-z0-9]+(?=,)
```

Sirve para extraer el usuario: toma desde el inicio y se detiene antes de la coma.

Idea clave: lookbehind mira hacia atras, lookahead mira hacia adelante. Ambos permiten extraer limpio, sin arrastrar el separador.

### Consumir vs no consumir (ejemplo minimo)

Texto: `ana,clave123`

- Regex que consume: `^[^,]*`
  - Resultado: `ana`
  - Explicacion: tomo caracteres reales del texto.

- Regex que no consume: `(?<=,)`
  - Resultado: vacio (no devuelve texto por si sola)
  - Explicacion: solo verifica que antes haya una coma.

Por eso muchas regex combinan ambas ideas: una parte ubica (no consume) y otra parte extrae (consume).

## Tema 5: guardar el primero y pedir el mismo al final

Patron:

```regex
(?<=,)(.).*\1$
```

Como leerlo facil:

1. `(?<=,)`: empezar despues de la coma (o sea, en la contraseña).
2. `(.)`: guardar el primer caracter de la contraseña.
3. `.*`: permitir cualquier cosa en el medio.
4. `\1`: exigir que el ultimo caracter sea igual al primero guardado.
5. `$`: terminar al final de linea.

Ejemplos rapidos:

- `unicornu` -> SI (empieza con `u` y termina con `u`).
- `BananaSplit99` -> NO (empieza con `B` y termina con `9`).

Idea clave: este patron sirve para comparar "primero" contra "ultimo" sin escribir una regex distinta para cada letra.

## Tema 6: validaciones compuestas

### Longitud exacta

```regex
(?<=,).{14}$
```

- `(?<=,)`: empezar despues de la coma.
- `.{14}`: exactamente 14 caracteres.
- `$`: fin de linea.

### Reglas multiples de seguridad

```regex
(?<=,)(?=.*[a-z])(?=.*[A-Z])(?=.*\d)(?=.*[@$!%*?&#]).+$
```

Cada lookahead agrega una regla:

- al menos una minuscula.
- al menos una mayuscula.
- al menos un numero.
- al menos un simbolo especial del conjunto indicado.

Principio general: cuando una validacion tiene muchas condiciones simultaneas, conviene separar cada regla en un lookahead.

## Resumen para estudiar rapido

Checklist de lectura de cualquier regex:

1. Inicio: `^`, `\b`, `(?<=...)`.
2. Cuerpo: literales, clases (`[A-Z]`, `[^,]`, `\d`, `\w`).
3. Cantidad: `*`, `+`, `{n}`.
4. Cierre: `$`, `(?=...)`, `\1`.

Si una regex falla, revisar ese orden suele mostrar rapido en que parte esta el error.
